# Mastering Pseudocode for Python
Welcome! This lesson will teach you how to bridge the gap between human logic and computational logic. 

**Pseudocode** is a detailed, step-by-step description of what a computer program must do, written in plain English rather than a strict programming language. Think of it as a blueprint for your code.

### Why write pseudocode?
1. **Reduces cognitive load:** Separate *what* you want to solve from *how* to type the syntax.
2. **Catches bugs early:** It is much easier to fix a logical flaw in plain English than in a 100-line script.
3. **Enhances communication:** It lets you share ideas with non-programmers or developers who use different languages.

### Core Rules for Writing Pseudocode
* Write **one statement per line**.
* Capitalize key structural commands (e.g., `IF`, `THEN`, `FOR`, `WHILE`, `OUTPUT`).
* Use **indentation** to show blocks of logic, just like Python.
* Keep it independent of language, but lean toward Python-like structures if that is your target language.

## Variables, Input, and Output
In pseudocode, we use descriptive terms to assign values and show data interacting with the user.

* **Pseudocode standard:** 
  ```text
  INPUT user_age
  SET threshold TO 18
  OUTPUT "Age recorded."
  ```

Let's see how a simple input/output routine translates directly from Pseudocode into Python code.

````python
# --- EXERCISE 1: Variable and Output Translation ---
# Read the pseudocode inside these comments, then run the cell to see the Python equivalent.

# PSEUDOCODE:
# SET user_name TO "Alex"
# OUTPUT "Hello, " + user_name

# PYTHON EQUIVALENT:
user_name = "Alex"
print(f"Hello, {user_name}")
````

## Conditional Logic (`IF/THEN/ELSE`)
Conditional logic helps your program make decisions based on rules. 

* **Pseudocode standard:**
  ```text
  IF score is greater than 50 THEN
      OUTPUT "Pass"
  ELSE
      OUTPUT "Fail"
  ENDIF
  ```

Notice that we use an explicit `ENDIF` in standard pseudocode to show exactly where the decision block finishes. Python handles this structurally using strict indentation.

````python
# --- EXERCISE 2: Conditional Logic ---
# Read the pseudocode and run the cell to see it working in Python.

# PSEUDOCODE:
# SET temperature TO 32
# IF temperature IS LESS THAN OR EQUAL TO 32 THEN
#     OUTPUT "It is freezing!"
# ELSE
#     OUTPUT "It is above freezing."
# ENDIF

# PYTHON EQUIVALENT:
temperature = 32

if temperature <= 32:
    print("It is freezing!")
else:
    print("It is above freezing.")
````

## Loops (`FOR` and `WHILE`)
Loops repeat actions until a condition is met. 

* **`FOR` loops** are used when you know exactly how many times you want to repeat something.
* **`WHILE` loops** are used when you want to repeat something until a condition changes.

```text
FOR each number from 1 to 5
    OUTPUT number
ENDFOR
```

````python
# --- EXERCISE 3: Iteration via Loops ---
# Read the pseudocode and run the cell to see the loop execute.

# PSEUDOCODE:
# FOR each item in list
#     OUTPUT item * 2
# ENDFOR

# PYTHON EQUIVALENT:
numbers = [1, 2, 3, 4, 5]
for item in numbers:
    print(item * 2)
````
